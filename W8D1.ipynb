## 🌟 Exercise 1: Open Source Levels Reflection

### Comparative Paragraph

**Fully Open** models provide access to the complete model architecture, pre-trained weights, and training code, allowing for maximum flexibility in modification and retraining. Their main advantage is complete control and transparency, enabling deep customization. **Weights Released** models offer pre-trained weights but not necessarily the full training code, meaning they are suitable for fine-tuning on specific tasks but limit architectural changes. The primary limitation here is the inability to fully understand or alter the training process. **Architecture Only** models reveal the model's structure but lack pre-trained weights, requiring users to train from scratch. While this offers architectural insights, its main disadvantage is the significant computational cost and time required for initial training.

### Healthcare Prompt Answer

For building a healthcare-specific assistant that must be retrained on clinical data, a **Fully Open** model is essential. This level supports domain-specific fine-tuning because it provides access to the complete training pipeline, allowing for rigorous customization and validation against sensitive healthcare data while ensuring full auditability and compliance.

## 🌟 Exercise 4: Local Readiness Audit

```python
import subprocess
import re

# Gather RAM info
try:
    ram_output = subprocess.check_output(['free', '-h']).decode('utf-8')
    total_ram_match = re.search(r'Mem:\s+(\d+\.?\d*G)', ram_output)
    total_ram = total_ram_match.group(1) if total_ram_match else 'N/A'
except FileNotFoundError:
    total_ram = 'N/A (free command not found)'

# Gather Disk Space info
try:
    disk_output = subprocess.check_output(['df', '-h', '/']).decode('utf-8')
    free_disk_match = re.search(r'/dev/root\s+\S+\s+\S+\s+(\d+\.?\d*G)', disk_output)
    free_disk = free_disk_match.group(1) if free_disk_match else 'N/A'
except FileNotFoundError:
    free_disk = 'N/A (df command not found)'

# Gather OS info
try:
    os_output = subprocess.check_output(['uname', '-a']).decode('utf-8')
    if 'Linux' in os_output:
        if 'WSL2' in os_output or 'microsoft' in os_output.lower():
            os_type = 'Linux (WSL2)'
        else:
            os_type = 'Linux'
    else:
        os_type = 'Other OS'
except FileNotFoundError:
    os_type = 'N/A (uname command not found)'

print(f"Total RAM: {total_ram}")
print(f"Free Disk Space (root): {free_disk}")
print(f"OS Type: {os_type}")
```

## 🌟 Exercise 6: Cloud vs. Local Deployment Plan

### Pros & Cons List (5 bullets total):

**Local Deployment**
*   ✔️ **Pros:**
    *   Low latency: No network overhead, ideal for real-time applications.
    *   Full control: Complete ownership over data, security, and hardware configuration.
    *   Cost-effective for long-term, constant use (after initial hardware investment).
*   ❌ **Cons:**
    *   High upfront hardware cost: Requires significant investment in GPUs and specialized hardware.
    *   Limited scalability: Scaling up requires buying more physical hardware, which is slow and expensive.

**Cloud Deployment**
*   ✔️ **Pros:**
    *   Easy scalability: Quickly provision or de-provision resources based on demand.
    *   Lower upfront cost: Pay-as-you-go model, no large initial hardware investment.
    *   Managed infrastructure: Cloud providers handle maintenance, security, and updates.
*   ❌ **Cons:**
    *   Potential for higher long-term costs: Can become expensive with continuous, heavy usage.
    *   Data privacy/security concerns: Data leaves local control, relying on provider's security measures.
    *   Latency due to network communication.

```python
# Optional Colab Report: Run a 7B Model on Colab

!pip install transformers accelerate

from transformers import AutoModelForCausalLM, AutoTokenizer
import time

# Due to potential resource limitations in a free Colab environment,
# using a smaller, more accessible 7B model or its quantized version is advisable.
# Llama-2-7b-chat-hf requires acceptance of its license on Hugging Face.
# For demonstration, we'll try a model that is generally easier to load without specific access tokens or very high RAM.
# Note: This might still push free Colab limits or require a GPU runtime.

model_name = "mistralai/Mistral-7B-Instruct-v0.2" # A good performing 7B model
tokenizer = AutoTokenizer.from_pretrained(model_name)

# device_map="auto" attempts to distribute the model across available devices (CPU/GPU)
# torch_dtype="auto" tries to use an appropriate data type (e.g., float16) to save memory

# It's important to set a trust_remote_code=True for some models, but generally for HuggingFace trusted models, it's not strictly needed.
# If you encounter CUDA out of memory errors, consider loading in 8-bit or 4-bit quantization:
# model = AutoModelForCausalLM.from_pretrained(model_name, device_map="auto", load_in_8bit=True)

try:
    print(f"Loading model: {model_name}...")
    model = AutoModelForCausalLM.from_pretrained(model_name, device_map="auto", torch_dtype="auto")
    print("Model loaded successfully.")

    inputs = tokenizer("Hello, world! Write a short poem about AI.", return_tensors="pt").to(model.device)

    print("Generating response...")
    start = time.time()
    outputs = model.generate(**inputs, max_new_tokens=50, do_sample=True, top_p=0.9, temperature=0.7)
    elapsed_time = time.time() - start

    print(tokenizer.decode(outputs[0], skip_special_tokens=True))
    print(f"\nElapsed: {elapsed_time:.2f} seconds")

except Exception as e:
    print(f"An error occurred during model loading or generation: {e}")
    print("This might be due to insufficient GPU RAM in your Colab session. "
          "Consider trying a smaller model, a quantized version, or upgrading to a Colab Pro session.")


```
