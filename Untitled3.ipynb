# US Superstore Data Analysis

## 1. Data Preparation

First, we need to load the US Superstore data. Please ensure you have uploaded the `US Superstore data.csv` file to your Colab environment or update the path in the code below if it's located elsewhere.

```python
import pandas as pd

# Load the dataset
try:
    df = pd.read_csv('US Superstore data.csv', encoding='latin1')
    print("Dataset loaded successfully!")
except FileNotFoundError:
    print("Error: 'US Superstore data.csv' not found. Please upload the file or check the path.")
    # Create an empty DataFrame or exit if the file is crucial
    df = pd.DataFrame()
except Exception as e:
    print(f"An error occurred while loading the dataset: {e}")
    df = pd.DataFrame()

# Display the first 5 rows of the DataFrame
if not df.empty:
    display(df.head())
```

```python
# Display basic information about the DataFrame
if not df.empty:
    print("\nDataFrame Info:")
    df.info()

    print("\nMissing values per column:")
    display(df.isnull().sum())
```

### Basic Data Cleaning and Preprocessing

We'll convert the 'Order Date' column to datetime objects and 'Sales' and 'Profit' to numeric types to ensure they are ready for analysis and visualization. We'll also handle any potential non-numeric values by coercing them to `NaN`.

```python
# Convert 'Order Date' to datetime
if 'Order Date' in df.columns:
    df['Order Date'] = pd.to_datetime(df['Order Date'], errors='coerce')
else:
    print("Warning: 'Order Date' column not found.")

# Convert 'Sales' to numeric, coercing errors
if 'Sales' in df.columns:
    df['Sales'] = pd.to_numeric(df['Sales'], errors='coerce')
else:
    print("Warning: 'Sales' column not found.")

# Convert 'Profit' to numeric, coercing errors
if 'Profit' in df.columns:
    df['Profit'] = pd.to_numeric(df['Profit'], errors='coerce')
else:
    print("Warning: 'Profit' column not found.")

# Convert 'Discount' to numeric, coercing errors
if 'Discount' in df.columns:
    df['Discount'] = pd.to_numeric(df['Discount'], errors='coerce')
else:
    print("Warning: 'Discount' column not found.")

# Drop rows with NaN values that resulted from coercion in critical columns
df.dropna(subset=['Order Date', 'Sales', 'Profit', 'Discount'], inplace=True)

print("Data cleaning complete. Displaying info after cleaning:")
df.info()
```

## 2. Data Visualization with Matplotlib

### Interactive Line Chart: Sales Trends Over the Years

We will create a line chart to visualize how sales have changed over time. For interactivity, in a local Jupyter environment, you might use `%matplotlib notebook` or `%matplotlib widget`. In Google Colab, basic interactive features like zooming and panning are usually available by default with `matplotlib.pyplot` plots.

```python
import matplotlib.pyplot as plt

# Ensure 'Order Date' is datetime before extracting year
if 'Order Date' in df.columns and pd.api.types.is_datetime64_any_dtype(df['Order Date']):
    df['Order Year'] = df['Order Date'].dt.year

    # Group by year and sum sales
    yearly_sales = df.groupby('Order Year')['Sales'].sum().reset_index()

    plt.figure(figsize=(12, 6))
    plt.plot(yearly_sales['Order Year'], yearly_sales['Sales'], marker='o', linestyle='-')
    plt.title('Sales Trends Over the Years')
    plt.xlabel('Year')
    plt.ylabel('Total Sales')
    plt.grid(True)
    plt.xticks(yearly_sales['Order Year'].unique()) # Ensure all years are shown as ticks
    plt.tight_layout()
    plt.show()
else:
    print("Cannot create sales trend plot: 'Order Date' column not found or not in datetime format.")
```

### Interactive Map: Sales Distribution by Country/State

The 'US Superstore' dataset primarily contains data for the United States. Therefore, instead of a global country map, we will visualize sales distribution by 'State' within the US.

Creating a truly 'interactive map' with geographical shapes directly within Matplotlib can be complex and typically requires external shapefile data (`geopandas` and `matplotlib.basemap` or `cartopy`). For simplicity and to fit within typical Colab setup, we will create a bar chart showing sales per state, which effectively visualizes the distribution. If a more advanced geographical map is strictly required, libraries like `folium` would be more appropriate for interactivity, but they fall outside the direct Matplotlib scope.

```python
# Check if 'Country' column exists and if it's mostly 'United States'
if 'Country' in df.columns and df['Country'].nunique() == 1 and df['Country'].iloc[0] == 'United States':
    print("Dataset is primarily for the United States. Visualizing sales by State.")
    if 'State' in df.columns:
        sales_by_state = df.groupby('State')['Sales'].sum().sort_values(ascending=False).reset_index()

        plt.figure(figsize=(15, 8))
        plt.bar(sales_by_state['State'], sales_by_state['Sales'])
        plt.title('Sales Distribution by State (United States)')
        plt.xlabel('State')
        plt.ylabel('Total Sales')
        plt.xticks(rotation=90) # Rotate state names for better readability
        plt.tight_layout()
        plt.show()
    else:
        print("Warning: 'State' column not found, cannot visualize sales distribution by state.")
elif 'Country' in df.columns:
    print("Visualizing sales by Country.")
    sales_by_country = df.groupby('Country')['Sales'].sum().sort_values(ascending=False).reset_index()

    plt.figure(figsize=(12, 6))
    plt.bar(sales_by_country['Country'], sales_by_country['Sales'])
    plt.title('Sales Distribution by Country')
    plt.xlabel('Country')
    plt.ylabel('Total Sales')
    plt.xticks(rotation=45) # Rotate country names if necessary
    plt.tight_layout()
    plt.show()
else:
    print("Cannot visualize sales distribution by country/state: 'Country' or 'State' column not found.")
```

## 3. Data Visualization with Seaborn

### Bar Chart: Top 10 Products by Sales

We will use Seaborn to create a bar chart showing the top 10 products with the highest sales.

```python
import seaborn as sns

if 'Product Name' in df.columns and 'Sales' in df.columns:
    top_10_products = df.groupby('Product Name')['Sales'].sum().nlargest(10).reset_index()

    plt.figure(figsize=(14, 7))
    sns.barplot(x='Sales', y='Product Name', data=top_10_products, palette='viridis')
    plt.title('Top 10 Products by Sales')
    plt.xlabel('Total Sales')
    plt.ylabel('Product Name')
    plt.tight_layout()
    plt.show()
else:
    print("Cannot create Top 10 Products by Sales plot: 'Product Name' or 'Sales' column not found.")
```

### Scatter Plot: Relationship Between Profit and Discount

We'll use a scatter plot to analyze how profit is affected by the discount applied to products.

```python
if 'Profit' in df.columns and 'Discount' in df.columns:
    plt.figure(figsize=(10, 6))
    sns.scatterplot(x='Discount', y='Profit', data=df, alpha=0.6, hue='Sales', size='Sales', sizes=(20, 400), palette='coolwarm')
    plt.title('Relationship Between Profit and Discount')
    plt.xlabel('Discount')
    plt.ylabel('Profit')
    plt.grid(True, linestyle='--', alpha=0.7)
    plt.tight_layout()
    plt.show()
else:
    print("Cannot create Profit vs. Discount scatter plot: 'Profit' or 'Discount' column not found.")
```

## 4. Comparative Analysis: Matplotlib vs. Seaborn

**Matplotlib:**
*   **Ease of Use:** Matplotlib provides fine-grained control over every aspect of a plot. This offers immense flexibility but can require more lines of code for complex customizations.
*   **Effectiveness:** Excellent for creating highly customized, publication-quality static plots. For basic interactive features in Colab, it often integrates well, allowing zoom/pan. However, for advanced interactive dashboards, other libraries might be preferred.
*   **Observations:** The line chart for sales trends and the bar chart for sales distribution were straightforward to create. Customizing labels, titles, and grid lines was intuitive.

**Seaborn:**
*   **Ease of Use:** Seaborn is built on Matplotlib and provides a high-level interface for drawing attractive statistical graphics. It generally requires less code to produce visually appealing and informative plots, especially for common statistical visualizations.
*   **Effectiveness:** Highly effective for exploratory data analysis, making it easy to visualize relationships between variables, distributions, and comparisons across categories with minimal code. Its default aesthetics are often more pleasing than Matplotlib's.
*   **Observations:** Creating the bar chart for top products and the scatter plot for profit vs. discount was very efficient. Seaborn handled grouping and aggregation implicitly in some cases, and its aesthetic defaults greatly improved the plots' readability and visual appeal. The `hue` and `size` parameters in the scatter plot, for example, added multiple dimensions of information with ease.

**Overall Comparison:**
*   **When to use Matplotlib:** For ultimate control, custom layouts, and when you need to integrate plots into larger applications or have very specific formatting requirements not easily met by Seaborn.
*   **When to use Seaborn:** For quick, aesthetically pleasing, and statistically oriented plots. It's fantastic for EDA and when you want to focus on insights rather than intricate plot details. Seaborn often reduces boilerplate code.

Both libraries are powerful and often used together, with Seaborn being used for initial high-level plots and Matplotlib for further customization or for types of plots not natively supported by Seaborn.
