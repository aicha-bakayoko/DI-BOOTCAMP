## Bank Customer Churn Prediction Exercise

This notebook will guide you through the process of building a machine learning model to predict customer churn for a bank.

### Step 1: Data Loading and Initial Exploration

First, we need to load the dataset. Since no specific dataset file was provided, I'll assume we are working with a CSV file named `Bank_Customer_Churn_Dataset.csv`. If you have a different file, please replace the filename in the code below. For demonstration purposes, I will create a dummy dataset to simulate the structure described in the problem statement.

```python
import pandas as pd
import numpy as np

# Create a dummy dataset resembling the description
np.random.seed(42)
data = {
    'CustomerId': range(1, 1001),
    'Surname': [f'Customer_{i}' for i in range(1, 1001)],
    'CreditScore': np.random.randint(350, 850, 1000),
    'Geography': np.random.choice(['France', 'Spain', 'Germany'], 1000),
    'Gender': np.random.choice(['Male', 'Female'], 1000),
    'Age': np.random.randint(18, 92, 1000),
    'Tenure': np.random.randint(0, 10, 1000),
    'Balance': np.random.uniform(0, 250000, 1000),
    'NumOfProducts': np.random.randint(1, 4, 1000),
    'HasCrCard': np.random.choice([0, 1], 1000),
    'IsActiveMember': np.random.choice([0, 1], 1000),
    'EstimatedSalary': np.random.uniform(100, 200000, 1000),
    'Exited': np.random.choice([0, 1], 1000, p=[0.8, 0.2]) # 20% churn rate
}

df = pd.DataFrame(data)

# Display the first 5 rows of the dataset
display(df.head())
```

Now, let's get a summary of the DataFrame to understand its structure, data types, and identify any missing values.

```python
# Get information about the DataFrame
display(df.info())
```

Next, let's look at descriptive statistics for the numerical columns.

```python
# Display descriptive statistics
display(df.describe())
```

### Step 2: Data Preprocessing

Data preprocessing is crucial for preparing our data for machine learning models. This involves several steps:

1.  **Handling Missing Values**: We'll check for and address any missing data points.
2.  **Encoding Categorical Features**: Machine learning models typically require numerical input, so we'll convert categorical variables into a numerical format.
3.  **Feature Scaling (Optional, but often beneficial)**: We might scale numerical features to ensure that no single feature dominates the learning process due to its magnitude.

#### 2.1 Handling Missing Values

Let's check if there are any missing values in our dataset.

```python
# Check for missing values
display(df.isnull().sum())
```

Since our dummy dataset doesn't have missing values, we can proceed. If there were any, common strategies include:

*   **Imputation**: Filling missing values with the mean, median, mode, or a constant.
*   **Deletion**: Removing rows or columns with missing values (use with caution to avoid data loss).

#### 2.2 Encoding Categorical Features

Our dataset contains categorical features like `Geography` and `Gender`. We need to convert these into numerical representations. One-hot encoding is a suitable method for this, creating new binary columns for each category.

```python
# Identify categorical features
categorical_features = ['Geography', 'Gender']

# Apply one-hot encoding
df_encoded = pd.get_dummies(df, columns=categorical_features, drop_first=True)

# Display the first few rows of the encoded DataFrame
display(df_encoded.head())
```

We also need to drop irrelevant columns that won't contribute to the model, such as `CustomerId` and `Surname`.

```python
# Drop irrelevant columns
df_final = df_encoded.drop(columns=['CustomerId', 'Surname'])

display(df_final.head())
```
