## 🌟 Exercise 1: Rewrite and Optimize a Vague Prompt

### Tasks

#### 1. Identify 3 critical issues with this prompt:

Original Prompt: “Write something about productivity tips.”

1.  **Lack of Specificity/Context:** The prompt doesn't specify the target audience (e.g., students, professionals), the platform (LinkedIn), or the brand voice (e.g., professional, casual, innovative). Without this, the AI cannot tailor the content effectively.
2.  **Missing Call to Action/Purpose:** The prompt doesn't indicate the desired outcome (e.g., drive app downloads, increase engagement, educate). A LinkedIn post typically aims for more than just providing tips; it needs to connect with the audience and encourage interaction.
3.  **No Format or Length Constraints:** There are no guidelines on how the information should be presented (e.g., bullet points, narrative, short paragraphs) or how long it should be, which is crucial for social media content where conciseness and scannability are key.

#### 2. Rewrite the prompt using all three techniques:

**Rewritten Prompt:**

```text
Act as a marketing specialist for a productivity startup called FlowNest. Your goal is to create a highly engaging LinkedIn post that promotes our new focus app, FlowNest, by offering valuable productivity tips.

Give 3 actionable productivity tips that directly relate to how FlowNest can help users achieve better focus. Each tip should include a clear, concise explanation of its benefit.

Format the post using a professional yet inspiring tone. Use emojis to break up text and maintain engagement, and ensure it includes a strong call to action at the end, encouraging readers to download FlowNest. The total post should be between 150-250 words.

Use this format whenever we promote productivity tools on social media.
```

## 🌟 Exercise 2: Multi-Part Prompt for Quiz Generation

### Tasks

#### 1. Write a complete prompt that asks the model to:

**Complete Prompt:**

```text
Act as an experienced 7th-grade science teacher. You have just provided an article about volcanic eruptions to your students. Your task is to generate engaging educational content based on this article for 11-13 year olds.

First, summarize the provided article on volcanic eruptions in exactly 2 simple bullet points, using vocabulary appropriate for a 7th-grade reading level. Focus only on the main causes and effects of eruptions.

Second, create 3 distinct multiple-choice questions about the article. For each question:
- It should have 3 answer options: 1 correct answer and 2 plausible but incorrect distractors.
- The vocabulary and concepts used in the questions and answers must be suitable for an 11-13 year old.

Present the summary and the quiz questions in a friendly and encouraging tone. Format the entire output to be easily transferable to Google Slides, with clear headings for the summary and each question.
```

#### 2. Explain three key ways your version prevents irrelevant or low-quality output.

Compared to the vague prompt “Make a quiz for kids about this article,” my version prevents irrelevant or low-quality output in the following ways:

1.  **Role Prompting and Target Audience Definition:** By specifying "Act as an experienced 7th-grade science teacher" and explicitly stating "for 11-13 year olds," the prompt guides the AI to adopt an appropriate pedagogical approach, tone, and vocabulary, ensuring the content is age-appropriate and relevant to the educational context, rather than a generic "quiz for kids."
2.  **Structured Output and Constraints:** The prompt clearly defines the desired output structure: "exactly 2 simple bullet points" for the summary, and "3 distinct multiple-choice questions" each with "1 correct answer and 2 plausible but incorrect distractors." This prevents the AI from generating an arbitrary number of points or questions, ensuring the output meets specific requirements and is easy to use in Google Slides.
3.  **Content Focus and Tone Specification:** I explicitly guide the summary's content ("Focus only on the main causes and effects of eruptions") and specify the tone ("friendly and encouraging tone"). This prevents the AI from including extraneous information or adopting an inappropriate tone, ensuring the output is focused, concise, and suitable for the intended educational environment.

## 🌟 Exercise 4: Match Prompt to Purpose

### Tasks

#### 1. Write your prompt using the selected style.

**Selected Style: Functional**

**Prompt for Customer Support Chatbot:**

```text
Based on the customer's query, identify the product mentioned, extract the core issue, and provide a single, direct solution from the product's FAQ or support knowledge base. If the issue requires human intervention (e.g., order modification, technical troubleshooting beyond FAQs), escalate the query by generating a concise message for a support agent, including the customer's name, product, and a brief description of the non-FAQ issue. If the product is not recognized or the issue is unclear, ask a clarifying question about the product or issue.
```

#### 2. Identify 2 key features that make your prompt a match for the style.

1.  **Task Execution with Precision:** The prompt clearly outlines a sequence of precise, defined tasks for the AI to perform: "identify the product," "extract the core issue," "provide a single, direct solution," "escalate the query by generating a concise message," and "ask a clarifying question." Each task is a distinct function the chatbot needs to execute.
2.  **Conditional Logic and Specific Outputs:** The prompt incorporates conditional logic ("If the issue requires human intervention..." and "If the product is not recognized or the issue is unclear...") and specifies exact output requirements for each condition (e.g., "single, direct solution," "concise message for a support agent," "clarifying question"). This demonstrates a focus on precise task execution rather than exploration or open-ended conversation.

#### 3. Justify your choice: Why is this style better than the others for your specific case?

The **Functional** style is better than the others for a customer support chatbot at an e-commerce company for several reasons:

*   **Efficiency and Accuracy:** Customer support requires quick, accurate resolutions. A functional prompt ensures the chatbot directly addresses the problem, provides relevant information, or efficiently escalates, minimizing unnecessary back-and-forth which can frustrate customers.
*   **Consistency:** Functional prompts lead to more consistent responses, which is vital for maintaining brand quality and ensuring all customers receive similar levels of service for common issues.
*   **Reduced Hallucinations/Irrelevant Information:** By focusing on specific tasks and drawing from a defined knowledge base (FAQ, support knowledge base), a functional prompt reduces the likelihood of the chatbot generating creative but irrelevant or incorrect information, which could be detrimental in a support context.
*   **Scalability:** A functional approach allows for easier updates and maintenance of the chatbot's knowledge and response patterns as products or policies change, as the core tasks remain stable while the data they draw from can be updated.

## 🌟 Exercise 6: Hallucination Spotting and Mitigation

### Tasks

#### 1. Write a precise prompt instructing the model to avoid hallucinations and use only the provided text.

**Prompt:**

```text
Summarize the following peer-reviewed article on climate change and marine biodiversity. Your summary must exclusively use information present within the provided text. Do not introduce any outside knowledge, assumptions, or claims that are not explicitly stated in the article.

[Insert Article Text Here]
```

#### 2. Rewrite the prompt using verification language, like:

**Rewritten Prompt:**

```text
Generate a summary of the provided peer-reviewed article concerning climate change and marine biodiversity. When constructing the summary, you must adhere to the following rules:

*   **Only include claims explicitly stated in the article.** Do not infer, assume, or add any information that cannot be directly verified by the text.
*   **If uncertain about a fact or unable to find direct support for a statement within the article, state that the information is not available or not covered in the provided text.** Do not invent information.
*   Present the key findings and conclusions as described solely by the article's content.

[Insert Article Text Here]
```

#### 3. List 2 mitigation strategies you used to reduce hallucination risk.

1.  **Constraint on Information Source (Exclusivity):** The prompt explicitly states to "exclusively use information present within the provided text" and "Do not introduce any outside knowledge, assumptions, or claims that are not explicitly stated in the article." This directly limits the AI's scope to the provided document, preventing it from drawing on its broader training data for potentially incorrect or unverified facts.
2.  **Verification Language and Uncertainty Handling:** By including directives like "Only include claims explicitly stated in the article" and "If uncertain, state that the information is not available," the prompt instructs the AI on how to handle missing or unverifiable information. This encourages the AI to either confirm a statement's presence in the text or admit its absence, rather than fabricating a response.

#### 4. Name 2 professional domains (e.g., legal, healthcare) where hallucinations are especially risky, and explain why.

1.  **Legal Domain:** Hallucinations are extremely risky in the legal domain because legal advice, case summaries, or document generation must be 100% accurate and based on established laws, precedents, and facts. A hallucinated legal precedent or misquoted statute could lead to incorrect legal advice, wrongful convictions, financial penalties, or severe professional repercussions for lawyers and their clients. The trust in legal counsel relies on verifiable, factual information.
2.  **Healthcare/Medical Domain:** In healthcare, inaccurate information or fabricated medical advice (hallucinations) can have life-threatening consequences. If an AI provides incorrect diagnoses, recommends inappropriate treatments, or fabricates drug interactions, it could lead to patient harm, misdiagnosis, delayed proper care, or even death. Medical decisions require absolute precision and adherence to verified scientific and clinical data.
