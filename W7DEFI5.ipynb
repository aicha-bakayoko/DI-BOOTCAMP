## Step 1: Craft a Prompt That Controls Output Tone, Format, and Length

Here's a prompt you can use to address the requirements for Step 1. Copy and paste this into your Generative AI model and observe the output.

```
Rewrite the following policy in a friendly and clear tone. Use bullet points to organize the information. Paraphrase the original text, avoiding any direct quotations. Ensure the total output is under 75 words.

Policy text:
"Employees must ensure that all remote access to internal systems is established via the approved secure VPN. Under no circumstances should unsecured connections or personal devices lacking endpoint protection be used to access proprietary data or sensitive communications."
```

## Step 2: Evaluate the Output

After generating the response from your Generative AI model using the prompt above, evaluate it against the following criteria:

*   **Relevance:** Does the output stick to the message of the source content?
*   **Clarity:** Is the message easier to understand for non-technical staff?
*   **Structure:** Is the information formatted into clear bullet points?
*   **Tone:** Is the language appropriate for internal employee communication?
*   **Length:** Does it stay under 75 words?
*   **Factual Accuracy:** Did the model make up or remove any critical security details?

Take note of any areas where the AI's output deviates from the instructions, especially concerning factual accuracy or the introduction of new information.

## Step 3: Detect and Mitigate Hallucinations

If your AI model's output from Step 1 included any information not present in the original policy text (hallucinations), you need to revise the prompt to prevent this. Here's a revised prompt that includes an instruction to strictly limit the model to the input content:

```
Rewrite the following policy in a friendly and clear tone. Use bullet points to organize the information. Paraphrase the original text, avoiding any direct quotations. Ensure the total output is under 75 words. Only paraphrase the content in the input text; do not add new recommendations or technologies.

Policy text:
"Employees must ensure that all remote access to internal systems is established via the approved secure VPN. Under no circumstances should unsecured connections or personal devices lacking endpoint protection be used to access proprietary data or sensitive communications."
```

## Step 4: Paraphrasing Deep Dive

Now, let's create a prompt specifically for a junior intern audience, focusing on plain language and short phrases. This prompt uses different constraints for the output structure and tone.

```
Paraphrase the following policy paragraph for a junior intern audience. Use plain language and short phrases. Use no more than 4 bullet points. Avoid any corporate or legal jargon. Keep the tone supportive and informative.

Policy text:
"Employees must ensure that all remote access to internal systems is established via the approved secure VPN. Under no circumstances should unsecured connections or personal devices lacking endpoint protection be used to access proprietary data or sensitive communications."
```

## Step 5: Quote Extraction Variant

Sometimes, extracting a direct quote is more effective. Here's a prompt to do just that:

```
From the following policy text, extract one direct quote that best captures the core security policy.

Policy text:
"Employees must ensure that all remote access to internal systems is established via the approved secure VPN. Under no circumstances should unsecured connections or personal devices lacking endpoint protection be used to access proprietary data or sensitive communications."
```

### Reflection Questions:

1.  **In what kind of internal communication would quoting be more appropriate than paraphrasing?**
    *   Quoting is more appropriate when the exact wording is crucial for legal compliance, regulatory requirements, or when conveying official statements where precision is paramount. For example, in formal policy documents, legal notices, safety instructions, or when citing specific terms and conditions.

2.  **When might quoting pose a risk?**
    *   Quoting can pose a risk if the original text is overly technical, jargon-filled, or easily misinterpreted by the audience. It can also be risky if the quote is taken out of context, leading to misunderstandings, or if the tone of the original text is not suitable for the communication channel, potentially alienating the audience. Additionally, long quotes can overwhelm readers and hinder clarity if not carefully selected.
